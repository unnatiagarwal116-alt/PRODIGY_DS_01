# Task 01 - Data Visualization

This project visualizes the population distribution of the top 10 countries in 2022 using a bar chart.

## Objective
Create a bar chart or histogram to visualize the distribution of a categorical or continuous variable.

## Dataset
World Bank population dataset.

## Tools Used
- Python
- Pandas
- Matplotlib
- Seaborn

## Conclusion
The chart shows that India and China had the highest populations in 2022, followed by the United States, Indonesia, Pakistan, Nigeria, Brazil, Bangladesh, Russian Federation, and Mexico.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv", skiprows=4)

year = "2022"

population_data = df[["Country Name", "Country Code", year]].dropna()

# Keep only actual countries by manually selecting top population countries
actual_countries = [
    "India",
    "China",
    "United States",
    "Indonesia",
    "Pakistan",
    "Nigeria",
    "Brazil",
    "Bangladesh",
    "Russian Federation",
    "Mexico"
]

top_10 = population_data[population_data["Country Name"].isin(actual_countries)]
top_10 = top_10.sort_values(by=year, ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(x=year, y="Country Name", data=top_10)

plt.title("Top 10 Countries by Population in 2022")
plt.xlabel("Population")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```
